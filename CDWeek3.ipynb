```python
%%writefile employee.l
%{
#include <stdio.h>
int valid = 0, invalid = 0;
%}

%%
[A-Z]{2}_?[0-9]{3}[a-z]* {
    printf("Valid Employee ID : %s\n", yytext);
    valid++;
}

[^ \t\n]+ {
    printf("Invalid Employee ID : %s\n", yytext);
    invalid++;
}

[ \t]+ ;
\n ;
%%

int yywrap()
{
    return 1;
}

int main()
{
    yylex();

    printf("Valid IDs : %d\n", valid);
    printf("Invalid IDs : %d\n", invalid);

    return 0;
}
```

```python
%%writefile token_analyzer.l
%{
#include <stdio.h>
int keywords = 0, identifiers = 0, numbers = 0;
int operators = 0, symbols = 0, lines = 1; // Start lines count at 1
%}

%%
"int"|"float"|"if"|"else"|"while"|"return" {
    keywords++;
}

[a-zA-Z][a-zA-Z0-9]* {
    identifiers++;
}

[0-9]+ {
    numbers++;
}

"+"|"-"|"*"|"/"|"=" {
    operators++;
}

[;,(){}] {
    symbols++;
}

\n {
    lines++;
}

[ \t]+ ;
. ;
%%

int yywrap()
{
    return 1;
}

int main()
{
    yylex();

    printf("Keywords : %d\n", keywords);
    printf("Identifiers : %d\n", identifiers);
    printf("Numbers : %d\n", numbers);
    printf("Operators : %d\n", operators);
    printf("Special Symbols : %d\n", symbols);
    printf("Lines : %d\n", lines);

    return 0;
}
```

```python
!lex token_analyzer.l
!gcc lex.yy.c -o token_analyzer
# Test with a sample C code snippet
!echo -e "int x = 10;\nfloat y = 20.5;\nif (x > y) { return 1; }" | ./token_analyzer
```

```python
%%writefile student_records.l
%{
#include <stdio.h>
int total = 0, valid = 0, invalid = 0;
int pass = 0, fail = 0;
int roll, marks;
char subject[10];
%}

%%
[0-9]{3}[ \t]+[A-Z]{2}[ \t]+[0-9]+ {
    total++;
    sscanf(yytext, "%d %s %d", &roll, subject, &marks);

    if (marks >= 0 && marks <= 100) {
        valid++;

        if (marks >= 40) {
            printf("Valid Record - PASS : %s", yytext);
            pass++;
        }
        else {
            printf("Valid Record - FAIL : %s", yytext);
            fail++;
        }
    }
    else {
        printf("Invalid Record (Marks out of range) : %s", yytext);
        invalid++;
    }
}

.* {
    total++;
    printf("Invalid Record : %s", yytext);
    invalid++;
}

\n ;

%%

int yywrap()
{
    return 1;
}

int main()
{
    yylex();

    printf("\n------ Student Marks Report ------\n");
    printf("Total Records : %d\n", total);
    printf("Valid Records : %d\n", valid);
    printf("Invalid Records : %d\n", invalid);
    printf("Passed : %d\n", pass);
    printf("Failed : %d\n", fail);

    return 0;
}
```

```python
!lex student_records.l
!gcc lex.yy.c -o student_records
# Test with valid, invalid, pass, fail, and out-of-range cases
!echo -e "101 CS 85\n102 MA 35\n103 PH 105\nInvalidLineHere\n104 CH 40" | ./student_records
```
