```python
%%writefile expression.l
%{
#include <stdio.h>
int op = 0, cp = 0, ob = 0, cb = 0, qm = 0;
%}

%%
"("     { op++; printf("TOKEN: (\n"); }
")"     { cp++; printf("TOKEN: )\n"); }
"["     { ob++; printf("TOKEN: [\n"); }
"]"     { cb++; printf("TOKEN: ]\n"); }
"?"     { qm++; printf("TOKEN: ?\n"); }
.       ;
\n      ;
%%

int main()
{
    printf("Enter the expression:\n");
    yylex();

    if (op == cp)
        printf("Parentheses are Balanced\n");
    else
        printf("Parentheses are Unbalanced\n");

    if (ob == cb)
        printf("Square Brackets are Balanced\n");
    else
        printf("Square Brackets are Unbalanced\n");

    printf("Total '?' operators found: %d\n", qm);

    return 0;
}

int yywrap()
{
    return 1;
}
```

```python
%%writefile ipv4.l
%{
#include <stdio.h>
#include <stdlib.h>
#include <string.h>

int valid(char *ip)
{
    int a, b, c, d;
    char extra;

    if (sscanf(ip, "%d.%d.%d.%d%c", &a, &b, &c, &d, &extra) != 4)
        return 0;

    if (a < 0 || a > 255 || b < 0 || b > 255 ||
        c < 0 || c > 255 || d < 0 || d > 255)
        return 0;

    char temp[100];
    sprintf(temp, "%d.%d.%d.%d", a, b, c, d);

    return strcmp(temp, ip) == 0;
}
%}

%%
[0-9]+(\.[0-9]+)* {
    if (valid(yytext))
        printf("Valid IPv4 Address: %s\n", yytext);
    else
        printf("Invalid IPv4 Address: %s\n", yytext);
}
[^ \t\n]+ {
    printf("Invalid IPv4 Address: %s\n", yytext);
}
[ \t]+ ;
\n ;
%%

int main()
{
    printf("Enter IP addresses:\n");
    yylex();
    return 0;
}

int yywrap()
{
    return 1;
}
```
